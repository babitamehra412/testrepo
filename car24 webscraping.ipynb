### WEB SCRAPING ###

*Extracting the data of brand Tata of used cars from Cars24.com of three different locations.*

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import html5lib
```

```python
url="https://www.cars24.com/buy-used-tata-cars-mumbai/"
```

```python
r= requests.get(url)
```

```python
response=r.status_code
response
```

```python
soup=BeautifulSoup(r.content,'html5lib')
```

```python
km_driven=[]
fuel_type=[]
transmission=[]
RTO=[]
j=soup.find_all('p',class_="sc-braxZu kvfdZL")
for i in j:
    if "km" in i.get_text():
        km=i.get_text()
        km_driven.append(km)
    elif "Diesel" in i.get_text() or "Petrol" in i.get_text() or "CNG" in i.get_text() or "Electric" in i.get_text():
        fuel=i.get_text()
        fuel_type.append(fuel)
    elif "Manual" in i.get_text() or "Auto" in i.get_text():
         t=i.get_text()
         transmission.append(t)
    elif "MH" in i.get_text():
         R=i.get_text()
         RTO.append(R)     
    
print(km_driven)
print(fuel_type)
print(transmission)
print(RTO)
print(len(km_driven),len(fuel_type),len(transmission),len(RTO))
```

```python
car_name=[]
names=soup.find_all('span',class_="sc-braxZu kjFjan")
for i in names:
    car_name.append(i.text)
print(len(car_name))
print(car_name)
```

```python
price=[]
prices=soup.find_all('p',class_="sc-braxZu cyPhJl")
for i in prices:
    price.append(i.text)
print(len(price))
print(price)  
```

```python
car_price=[]
car_price=price[2:22]
print(car_price)
print(len(car_price))
```

```python
location=[]
locations=soup.find_all('div',class_="styles_hubAddress__URioy")
for i in locations:
    location.append(i.text)
print(location)
print(len(location))    
```

```python
data=pd.DataFrame({'car_name':car_name,'price':car_price,'RTO':RTO,'km_driven':km_driven,'fuel_type':fuel_type,'transmission':transmission,'location':location})
data
```

```python
data.to_csv('cars24.csv')
```
